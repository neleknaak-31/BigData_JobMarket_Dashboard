# Q-Phase: Question & Quality Definition
## Projekttitel: Data-Driven Job Market Intelligence Dashboard für Deutschland

**Modul:** Big Data & Data Analytics  
**Vorgehensmodell:** QUA3CK (Phase Q)

---

### 1. Projektidee & Motivation
Im Rahmen dieser Q-Phase wird die Grundlage für eine interaktive Web-App auf Basis von **Streamlit** gelegt. Ziel ist es, den deutschen Arbeitsmarkt datenbasiert zu analysieren, Trends sichtbar zu machen und Handlungsempfehlungen abzuleiten.

**Fokus:** Tech- und Data-Berufe (hochgradig datengetrieben).

### 2. Problemstellung

Der deutsche Arbeitsmarkt, insbesondere im IT- und Technologiesektor, steht vor einer paradoxen Herausforderung: Trotz eines prognostizierten Fachkräftemangels von bis zu 725.000 unbesetzten Stellen in MINT-Berufen bis 2026 (IW Köln) fällt es vielen qualifizierten Jobsuchenden, Hochschulabsolventen und Quereinsteigern schwer, eine passende Anstellung zu finden.

Diese Diskrepanz entsteht durch drei zentrale Probleme, die dieses Projekt lösen soll:

1.  **Fehlende Markttransparenz (Markt-Ebene):** Informationen über Gehälter, regionale Nachfrage und notwendige Qualifikationen sind fragmentiert, oft veraltet oder basieren auf intransparenten Algorithmen privater Jobportale. Es fehlt an einer unabhängigen, datengestützten Gesamtübersicht auf Basis offizieller Open-Data-Quellen (wie Bundesagentur für Arbeit und Destatis).
2.  **Unklare Anforderungsprofile (Skill-Ebene):** Technologien im Tech-Sektor entwickeln sich rasand (z.B. GenAI). Stellenanzeigen sind oft überladen und spiegeln nicht die tatsächlichen Kernkompetenzen wider, die für eine Rolle (z.B. Data Scientist) essenziell sind. Jobsuchende wissen nicht, welche Skills sie *wirklich* priorisieren müssen.
3.  **Individueller "Skill Gap" (Nutzer-Ebene):** Nutzer kennen oft ihren "Ist-Zustand" (ihre aktuellen Skills), wissen aber nicht, wie groß die Lücke ("Gap") zum "Soll-Zustand" (den Marktanforderungen) ist. Es fehlen Werkzeuge, die diese Lücke personalisiert berechnen und konkrete Handlungsempfehlungen für die Weiterbildung geben.

**Die Kernfrage des Projekts lautet:**
*Wie kann eine interaktive Data-Analytics-Anwendung die Lücke zwischen individuellen Tech-Skills und realen Marktanforderungen in Deutschland transparent machen und Jobsuchende datengestützt bei der Karriereplanung unterstützen?*

### 3. Zielgruppe & Fragestellungen (The Question)
Um den "Business Value" sicherzustellen, definiert die Q-Phase die **Zielgruppe**:
*   **Jobsuchende & Studierende:** Orientierung im Tech-Markt.
*   **Quereinsteiger:** Identifikation von notwendigen Weiterbildungen.
*   **HR-Analysten:** Beobachtung von Markt-Trends.

**Zentrale Forschungsfragen:**
* Welche Berufe sind in Deutschland aktuell besonders gefragt?  
* Welche Skills werden am häufigsten benötigt?  
* Wie unterscheiden sich Gehälter je nach Rolle und Erfahrung?  
* Welche spezifischen Fähigkeiten fehlen für eine Ziel-Rolle?

### 4. Funktionale Anforderungen (App-Scope)
Die App geht über ein klassisches Dashboard hinaus durch:

*   **3.1 Interaktives Dashboard:** Filter nach Jobrollen, Region und Erfahrungslevel.
*   **3.2 Arbeitsmarktanalyse:** Identifikation von Engpassberufen und regionalen Unterschieden.
*   **3.3 Skill Demand Analyse:** Ranking der meistgefragten Technologien (Kombination von Arbeitsmarktdaten & Umfragen).
*   **3.4 Gehaltsanalyse:** Durchschnittsgehälter im deutschen Kontext.
*   **3.5 Skill Gap Analyzer (Zentrale Innovation):** Abgleich von Nutzer-Skills mit Marktanforderungen.
*   **3.6 Karriere-Empfehlungssystem:** Datengestützte Vorschläge für passende Jobs und Lernpfade.

### 5. Datenquellen & Qualitätssicherung (Quality)
Die Validität der Ergebnisse wird durch die Kombination seriöser Datenquellen gewährleistet:

1.  **Bundesagentur für Arbeit:** Offizielle Engpassanalysen & Nachfrage.
2.  **Destatis:** Validierte Einkommensstrukturen.
3.  **Eurostat:** Europäischer Vergleichskontext.
4.  **Stack Overflow:** Spezifische Tech-Skill-Trends.

**Qualitätsprüfung:** In der Q-Phase stellen wir sicher, dass die Daten aktuell sind (2024-2026) und eine ausreichende Granularität für den deutschen Markt (Bundesland-Ebene) besitzen.

### 6. Erfolgsmetriken (KPIs) der Q-Phase

Um den Erfolg des "Job Market Intelligence Dashboards" objektiv zu bewerten, wurden folgende Zielmetriken definiert:

#### A. Analyse- & Vorhersagequalität (Accuracy)
*   **Skill-Matching Accuracy (> 90%):** Die Genauigkeit, mit der eingegebene Nutzer-Skills korrekt den Marktanforderungen der Zielrolle zugeordnet werden (validiert durch einen manuellen Test-Datensatz).
*   **Salary Deviation (< 15%):** Die Abweichung der berechneten Gehaltsprognosen im Vergleich zu den offiziellen Median-Entgelten der Bundesagentur für Arbeit.

#### B. System-Performance & Deployment
*   **Response Time (< 800 ms):** Die Zeit, die das System benötigt, um nach der Eingabe der Skills die "Gap-Analyse" grafisch aufzubereiten.
*   **Deployment:** Erfolgreiches Hosting der Anwendung als öffentliche **Streamlit Web-App** via GitHub, die ohne lokale Installation im Browser aufrufbar ist.

#### C. Inhalts- & Datenqualität
*   **Data Freshness (< 1 Jahr):** Sicherstellung, dass kein Datensatz älter als 12 Monate ist, um die Aktualität von Tech-Trends (z. B. KI-Skills) zu garantieren.
*   **Coverage:** Abdeckung von mindestens **15 der relevantesten Tech-Jobprofile** (z. B. Data Scientist, Cloud Architect, Cyber Security Analyst) für den deutschen Markt.

### 7. Methodik & Mehrwert
Die Umsetzung erfolgt mittels **Python** (Pandas, Plotly) und **Streamlit**. 

**Innovativer Mehrwert:** 
Im Sinne von QUA3CK wird hier nicht nur visualisiert, sondern ein **Entscheidungsunterstützungssystem** geschaffen. Die Kombination aus behördlichen Makro-Daten und Community-Mikro-Daten ermöglicht eine Präzision, die herkömmliche Jobbörsen nicht bieten.
